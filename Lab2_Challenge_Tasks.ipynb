```python
import random

SURNAME = "Manansala"
SEED_NUM = 600
NUM_READINGS = 7
```

```python
def generate_sensor_data(surname, seed_num, count):
    """Generates sensor readings based on surname and SEED_NUM."""
    combined_seed = sum(ord(c) for c in surname) + seed_num
    random.seed(combined_seed)

    raw_inputs = []
    invalid_samples = ["ERR", "N/A", "999.9", "abc"]

    for _ in range(count):
        if random.random() < 0.2:
            raw_inputs.append(random.choice(invalid_samples))
        else:
            raw_inputs.append(str(round(random.uniform(-10.0, 110.0), 2)))
    return raw_inputs

def validate_and_classify(val_str):
    """Validates the input and classifies valid reading into ranges."""
    try:
        val = float(val_str)
        if 0.0 <= val <= 100.0:
            if val <= 30.0:
                cat = "Low"
            elif val <= 70.0:
                cat = "Normal"
            elif val <= 90.0:
                cat = "High"
            else:
                cat = "Critical"
            return True, cat, val
        else: 
            return False, "OUT_OF_RANGE", None
    except ValueError:
        return False, "INVALID_FORMAT", None
```

```python
def main():
    print("Assessment Data:")
    print(f"Surname: {SURNAME}")
    print(f"Seed Number: {SEED_NUM}")
    print(f"Total Readings: {NUM_READINGS}")
    
```

```python
raw_data = generate_sensor_data(SURNAME, SEED_NUM, NUM_READINGS)
print("Generated Sensor Data:")
print(raw_data)
print()

valid_results = []
invalid_results = []
classification_counts = {"Low": 0, "Normal": 0, "High": 0, "Critical": 0}
execution_logs = []

for idx, item in enumerate(raw_data, 1):
    is_valid, status, num_val = validate_and_classify(item)
    if is_valid:
        valid_results.append(num_val)
        classification_counts [status] += 1
        execution_logs.append(f"Reading {idx:02d} [{item}]: VALID -> Classified as '{status}'")
    else:
        invalid_results.append(item)
        execution_logs.append(f"Reading {idx:02d} [{item}]: INVALID -> Reason: {status}")
        
```

```python
print ("Valid/Invalid Results:")
print(f"Valid Readings ({len(valid_results)}): {valid_results}")
print(f"Invalid Readings ({len(invalid_results)}): {invalid_results}\n")
```

```python
print("Classification Results")
for cat, count in classification_counts.items():
    print(f" {cat:<10}: {count}")
print()
```

```python
print("Execution Log:")
for log in execution_logs:
    print(f" {log}")
print()
```

```python
print("Final Output:")
print(f"Total Input Items  : {len(raw_data)}")
print(f"Success Rate       : {(len(valid_results)/len(raw_data))*100:.1f}%")
if valid_results:
    print(f"Average Value      : {sum(valid_results)/len(valid_results):.2f}")
    print(f"Minimum Value      : {min(valid_results):.2f}")
    print(f"Maximum Value      : {max(valid_results):.2f}")

if __name__ == "__main__":
    main()
```

```python
def generate_signal(surname, seed_num, length):
    combined_seed = sum(ord(c) for c in surname) + seed_num
    random.seed(combined_seed)

    sample_pool = ['A', 'B', 'C', 'X', 'ERR', '12', '45', '88', '102', '-5']
    return [random.choices(sample_pool) for _ in range(length)]
def process_signal(raw_signal):
    processed = []
    for item in raw_signal:
        try:
            val = float(item)
            norm_val = round(max(0.0, min(1.0, val / 100.0)), 2)
            processed.append(norm_val)
        except ValueError:
            processed.append(item)
    return processed
def analyze_and_classify(processed_signal):
    char_analysis = {"alphabetic": 0, "numeric": 0, "invalid": 0}
    classifications = []
    logs = []

    for idx, element in enumerate(processed_signal, 1):
        str_elem = str(element)

    if str_elem.isalpha():
        char_analysis["alphabetic"] += 1
        category = "Control Character"
    elif isinstance(element, (int, float)):
        char_analysis["numeric"] += 1
        category = "High Signal" if element > 0.5 else "Low Signal"
    else: 
        char_analysis["invalid"] += 1
        category = "Noise / Error"

    classifications.append(category)
    logs.append(f"Step {idx:02d} [{element}]: Categorized as -> {category}")

    return char_analysis, classifications, logs
```

```python
def main():
    print("Assessment Data:")
    print(f"Surname: {SURNAME}")
    print(f"Seed Number: {SEED_NUM}")
    print(f"Signal Length: {NUM_READINGS}\n")
    
```

```python
raw_signal = generate_signal(SURNAME, SEED_NUM, NUM_READINGS)
print("Generated Signal:")
print(raw_signal)
print()

    
```

```python
def process_signal(raw_signal):
    processed = []
    for item in raw_signal:
        signal_value = item[0] if isinstance(item, (list, tuple)) and len(item) > 0 else item
        try:
            val = float(signal_value)
            norm_val = round(max(0.0, min(1.0, val / 100.0)), 2)
            processed.append(norm_val)
        except (TypeError, ValueError):
            processed.append(signal_value)
    return processed

proc_signal = process_signal(raw_signal)
print("Processed Signal:")
print(proc_signal)
print()
print("Processed Signal:")
print(proc_signal)
print()
```

```python
char_analysis, classifications, logs = analyze_and_classify(proc_signal)
print("Character Analysis:")
for key, count in char_analysis.items():
    print(f" - {key.capitalize()}: {count}")
print()
```

```python
print("Signal Classification:")
for idx, (sig, cls) in enumerate(zip(proc_signal, classifications), 1):
    print(f" Signal {idx:02d} ({sig}): {cls}")
print()
```

```python
print("Execution Log:")
for log in logs: 
    print(f" {log}")
print()
```

```python
print("Final Output:")
print(f"Total Elements Processed : {len(proc_signal)}")
print(f"Valid Numeric Signals    : {char_analysis['numeric']}")
print(f"Control Characters       : {char_analysis['alphabetic']}")
print(f"Flagged Noise            : {char_analysis['invalid']}")

if __name__ == "__main__":
    main()
```
